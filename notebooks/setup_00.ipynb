```python
import pandas as pd

path = "../data/raw/accepted_2007_to_2018Q4.csv.gz"


df = pd.read_csv(path, low_memory=False)
print(df.shape)

# The raw CSV has trailing footer/summary rows (e.g. "Total amount funded in
# policy code 1: ...") instead of real loan records, which breaks id's dtype.
df = df[df["id"].apply(lambda x: isinstance(x, int) or str(x).isdigit())].copy()
df["id"] = df["id"].astype("int64")
print(df.shape)

print(df.columns.tolist())
print(df.dtypes)
print(df.head())
print(df.describe())
df.to_parquet("../data/interim/accepted_2007_to_2018Q4.parquet", index=False)
```

# Understanding Week 0

Here we have a data set that is quite large. The set is ~2.26 million rows with 151 columns.
The raw CSV had a trailing footer/summery rows instead of real loan records. This broke id's dtype. These rows were dropped.

## Data transformation

The file was read compressed (.gz) and saved to interim data as `.parquet`.
A parquet stores data by column allowing reads to be faster.
